**Try:** move the parameter sliders; **Play** (each parameter sweeps from 0 to min, to max and back), **Stop** while it plays; **Reset**; **Variance** (plots the eigenvalue ratios); the *Deformation* / *Vectors* modes.

```python
%matplotlib inline
import warnings; warnings.simplefilter("ignore")
import numpy as np
import menpo.io as mio
from menpo.model import PCAModel
from menpo.shape import PointCloud
import menpowidgets as mw
print("menpowidgets", mw.__version__)
```

```python
takeo = mio.import_builtin_asset.takeo_ppm()
shape_model = PCAModel([PointCloud(np.random.RandomState(i).rand(20, 2) * 50) for i in range(5)])
app_images = [takeo.rescale(0.2).as_greyscale() for _ in range(6)]
for i, im in enumerate(app_images):
    im.pixels = im.pixels + 0.05 * np.random.RandomState(i).rand(*im.pixels.shape)
app_model = PCAModel(app_images)
```

## `visualize_shape_model_2d`

```python
mw.visualize_shape_model_2d(shape_model)
```

```python
mw.visualize_appearance_model(app_model)
```
